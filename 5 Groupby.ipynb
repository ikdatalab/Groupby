# 5 Groupby Method

```python
import pandas as pd
import numpy as np
```

## 1 Method

```python
data = pd.read_csv('order_data.csv')
data.head()
```

```python
data.groupby('customer_name')['order_total'].sum().sort_values(ascending=False)
```

## 2 Method

```python
data.head()
```

#### use (aggregate) - agg()

```python
data.groupby('customer_name').agg({'order_total' : ['count' , 'sum' , 'max' , 'min' , 'mean']}).round(2)
```

```python
all = data.groupby('customer_name').agg({'order_total' : ['count' , 'sum' , 'max' , 'min' , 'mean']}).round(2)
all.head(10)
```

## 3 Method

Haar 1 __payment_method__ ke through `kitna` sales hoa kea `averege` tha kitna `maximum` sale hoa aur `minimum` aur `total` with proper naming

```python
data.head()
```

```python
data.groupby('payment_method')['order_total'].agg(total_count='count',
                                                  total_revenue='sum',
                                                  max_value='max',
                                                  min_value='min',
                                                  ave_value='mean').round(2)
```

## 4 Method

tupple ke through groupby `new` data par

```python
product = pd.read_csv('products_data.csv')
product.head()
```

```python
product.groupby('brand').agg(total_order=('quantity','sum'),
                           total_sales=('sell_price','sum'),
                           max_sales=('sell_price','max'),
                           min_sales=('sell_price','min'),
                           ave_sales=('sell_price','mean'),
                           max_discount=('discount','max'),
                           ave_discount=('discount','mean')).round(2)
```
